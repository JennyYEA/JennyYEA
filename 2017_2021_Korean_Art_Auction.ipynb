```python
kart.groupby('Artist').agg(count_Artist = ('Artist', 'count')).sort_values('count_Artist', ascending = False)
```

```python
#2017-2021 한국 경매 낙찰 작품 수에 따른 작가 순위(작가 미상 제외 상위 10명)
kart.groupby('Artist').agg(count_Artist = ('Artist', 'count'))\
    .sort_values('count_Artist', ascending = False).head(11)
```

```python
#2017-2021 한국 경매 낙찰 금액 순위(상위 30 작품)
kart2 = kart.query('Artist in ["이우환","김환기","쿠사마 야요이","박서보","박수근","정상화","김창열","마르크 샤갈","천경자","이중섭"]')\
            .sort_values('Bidding', ascending = False).head(30)
```

```python
#2017-2021 한국 경매 낙찰 금액 순위 작가(상위 30 작품의 작가)
kart2.groupby('Artist').agg(count_Artist = ('Artist', 'count'))\
     .sort_values('count_Artist', ascending = False)
```
